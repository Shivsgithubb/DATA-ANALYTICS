```python
def even_odd(num):
    if num % 2 == 0:
        print("even")
    else:
        print("odd")
even_odd(10)
    

```

```python
#default parameter function
def myCountry(country = "india"):
    print("i am from " + country)

myCountry("usa")
myCountry("australia")
myCountry("")
myCountry("uk")
```

```python
def func(food):
    for x in food:
        print(x)
fruits = ['guava','pineapple','cherry']
func(fruits)
```

```python
evenodd = lambda num: 'even' if num%2==0 else "odd"
print(evenodd(10))
```

```python
add = lambda x,y : x+y
print(add(10,20))
```

```python
cube = lambda x,y: x**y

print(cube(5,3))
```
