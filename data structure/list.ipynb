```python
apps = ['netflix','chrome','google','songs','hotstar','swiggy','zomato','youtube','pictures']
print(apps)
```

```python
#positive indexing starts from 0
print(apps[5])
#negative indexing starts from -1 not from 0
print[-2]
```

```python
print(apps[2:5])
print(apps[2:9:2])
print(apps[2:])
print(apps[2::2])
print(apps[:2])
print(apps[:])
```

lists Methods 

- append - to add an element at the end of the list

- insert - to add an element at any specific position in the list

- pop - to delete an item using index position of the element

- remove - to delete an item using its Name

- copy - to make a copy of list

- sort - to sort item in any particular order

- extend - to combine two list in list

- reverse - to reverse the order of list

- clear - to remove all the item of the list but not the list

```python
#- append - to add an element at the end of the list

apps.append('snap')
print(apps)
```

```python
#- insert - to add an element at any specific position in the list

apps.insert(2,'ola')
print(apps)
```

```python
#- pop - to delete an item using index position of the element
apps.pop(4)
print(apps)
```

```python
#- remove - to delete an item using its Name
apps.remove("ola")
print(apps)
```

```python
#- copy - to make a copy of list
x = apps.copy()
print(x)
```

```python
x.clear()
print(x)
```

```python
#- sort - to sort item in any particular order
apps.sort()
print(apps)
```

```python
#- reverse - to reverse the order of list
apps.reverse()
print(apps)
```

```python
#- extend - to combine two list in list
apps2 = ['water','juice','drink']
apps.extend(apps2)
print(apps)
```

```python
apps2.extend(apps)
print(apps2)
```
