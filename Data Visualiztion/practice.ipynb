```python
# Histogram

import matplotlib.pyplot as plt

population_ages = [22,55,62,45,21,34,23,43,56,78,23,45,67,89,12,34,56]

# bins are the ranges of the histogram
bins=[10,20,30,40,50,60,70,80,90]

plt.hist(population_ages,bins, histtype='bar', rwidth=0.8, color='blue', alpha=0.7)
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('Histogram')
plt.show()
```

```python
# scatter plot

import matplotlib.pyplot as plt

x = [5,8,10,12,14,18,20]
y = [12,16,6,7,8,9,11]

plt.scatter(x,y, label='scatter plot', color='blue', s=30, marker='o')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Scatter plot')
plt.legend()
plt.show()
```

```python
# stack plot

import matplotlib.pyplot as plt

days = [1,2,3,4,5,6]

sleeping = [7,8,6,11,7,9]
eating = [2,3,4,3,2,4]
working = [7,8,6,11,7,9]
playing = [2,3,4,3,2,4]

plt.plot([],[],color='blue', label='sleeping', linewidth=5)
plt.plot([],[],color='green', label='eating', linewidth=5)
plt.plot([],[],color='red', label='working', linewidth=5)
plt.plot([],[],color='tan', label='playing', linewidth=5)

plt.stackplot(days, sleeping, eating, working, playing, colors=['blue','green','red','tan'])

plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('Stack plot')
plt.legend()
plt.show()
```

```python
# pie chart

slices = [7,2,2,13]
labels = ['a','b','c','d']
colors = ['blue','green','red','tan']

plt.pie(slices, labels=labels, colors=colors, startangle=90, shadow=True, explode=(0.1,0.1,0.1,0.1), autopct='%1.1f%%')

plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.capitalized

plt.title('Pie chart')
plt.show()
```
