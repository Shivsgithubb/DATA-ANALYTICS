```python
# Average tip by gender

avg_tip_by_gender = df.groupby('sex')['tip'].mean()
print("Average tip by gender:")
print(avg_tip_by_gender)
```

```python
# The day with the highest total bill collection
highest_bill_day = df.groupby('day')['total_bill'].sum().idxmax()
print(f"The day with the highest total bill collection: {highest_bill_day}")
```

```python

# Create a histogram of the total_bill column
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['total_bill'], bins=20, kde=False, color='blue')
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
# Create a boxplot showing the distribution of tip amounts for each day
sns.boxplot(data=df, x='day', y='tip', palette='coolwarm')
plt.title('Boxplot of Tip Amounts by Day')
plt.xlabel('Day')
plt.ylabel('Tip')
plt.show()
```

```python
# Create a violin plot to compare tip distribution across different sex categories
sns.violinplot(data=df, x='sex', y='tip', palette='muted')
plt.title('Violin Plot of Tip Distribution by Sex')
plt.xlabel('Sex')
plt.ylabel('Tip')
plt.show()
```

```python
# Create a scatter plot showing the relationship between total_bill and tip, colored by sex
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex', palette='coolwarm')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

```python
# Create an interactive bar chart showing average tip by day

avg_tip_by_day = df.groupby('day')['tip'].mean().reset_index()
fig = px.bar(avg_tip_by_day, x='day', y='tip', color='day', title='Average Tip by Day', 
             labels={'tip': 'Average Tip'}, barmode='group')
fig.show()
```

```python
# Interactive scatter plot of total_bill vs tip
fig = px.scatter(
    df,
    x='total_bill',
    y='tip',
    color='sex',
    size='size',
    title='Interactive Scatter Plot: Total Bill vs Tip',
    labels={'total_bill': 'Total Bill', 'tip': 'Tip'}
)
fig.show()
```

```python
# Interactive pie chart of the proportion of smokers vs non-smokers
smoker_counts = df['smoker'].value_counts()
fig = px.pie(
    names=smoker_counts.index,
    values=smoker_counts.values,
    title='Proportion of Smokers vs Non-Smokers',
    labels={'index': 'Smoker Status', 'values': 'Count'}
)
fig.show()
```
